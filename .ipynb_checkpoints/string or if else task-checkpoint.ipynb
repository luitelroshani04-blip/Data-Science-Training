```python
# Python program that creates a dictionary representing a student's grades in different subjects
# and then calculates the student's average grade

student_grade = {
    "English" : 90,
    "Math" : 85,
    "Science" : 80
}
total_marks = sum(student_grade.values())
avg = total_marks / len(student_grade)
print("Average Grade =", avg)





```

```python
#Task 4. 
#Write a program to input a character and check whether it is a vowel (a, e, i, o, u) or a consonant.
char= input("Enter a character").lower()
vowel = ['a', 'e', 'i', 'o', 'u']
if len(char) ==1:
    if char in vowel:
        print("Vowel")
    else:
        print("Consonant")
else:
    print("Invalid")
    
```

```python
#Task 13. Write a Python program to create a simple calculator that takes two numbers (num1 and num2) and an operator (+, −, ×, ÷) as input from the user. Perform the corresponding operation based on the given operator using elif statements. 
#Note: Handle division by zero appropriately.
a = float(input("Enter 1st num"))
b = float(input("Enter 2nd num"))
sign = input("select: +, −, ×, /")
if sign == "+":
    print(a+b)
elif sign == "-":
    print(a-b)
elif sign == "*":
    print(a*b)
elif sign == "/":
    if b==0:
        print("Undefined")
    else:
        print (a/b)
else:
    print("Invalid")

```

```python
#Task 14. 
#Write a program to input marks and assign grades based on the following conditions: 
'''• 90 and above → Grade A • 80 and above → Grade B • 70 and above → Grade C • 60 and above → Grade D • Below 60 → Fail'''
marks = float(input("Enter your marks"))
if marks >= 90:
    print("Grade A")
elif marks >= 80 and marks <90:
    print("Grade B")
elif marks >= 70 and marks <80:
    print("Grade C")
elif marks >= 60 and marks <70:
    print("Grade D")
else:
    print("Fail")
    
```

```python
#Task 17. 
#Write a program that takes three sides of a triangle and checks whether the sides can form a valid triangle.
#Note: A triangle is valid if the sum of any two sides is greater than the third side.
a = input("Enter 1st side of triangle")
b = input("Enter 2nd side of triangle")
c = input("Enter 3rd side of triangle")
if a+b > c:
    print("Valid triangle")
elif a+c > b:
    print ("valid triangle")
elif b+c > a:
    print ("valid tringle")
else:
    print("Invalid triangle")
```

```python
#Task 18. 
#Write a program that calculates an electricity bill based on units consumed:
#Units ≤ 100 : Rate = Rs. 5 per unit • Units 101 – 200 : Rate = Rs. 7 per unit • Units > 200 : Rate = Rs. 10 per unit
unit = float(input("Enter consumed unit"))
if unit <= 100:
    print("bill: Rs.", unit*5)
elif unit >= 101 and unit <= 200:
    print("bill: Rs.", unit*7)
else:
    print("bill: Rs.", unit*10)
```
