```python
#Take a number from the user and print numbers from 1 to that number.

num = int(input("Enter a number"))
for i in range (1,num+1):
    print(i)
          
```

```python
#Countdown
#Print numbers from 10 down to 1
for i in range (10,0,-1):
    print(i)
```

```python
# Task: Numbers Divisible by 3
# Print all numbers from 1 to 50 that are exactly divisible by 3.
# Use a for loop and the % operator.

for i in range (1,50):
    if i%3==0:
        print(i)
```
