```python
'''
for loop : used if no. of iteartion is known, used for sequence traversal

syntax:
for item in iterable:
    block of code

while loop : used if no. of iteration is unknown

'''
names = ["Roshani", "Sanjita", "Bhawana", "Kusum"]
for i in names:
    print("Hello", i)

```

```python
name = "sujan"
for i in name:
    print(i)
```

```python
#dictionary
a= {
    "name": "sujan",
    "age": 90
}
for i,j in a.items():
    print(i,j)
    
```

```python
# range
#syntax : range (start,end, step)
for i in range (0,10):
    print("sujan",i)
```

```python
#even number upto 20
for i in range(0,20,2):
    print(i)
```

```python
# 5,4,3,2,1
for i in range(5,0,-1):
    print(i)
```

```python
#multipliation table
num = int(input("Enter a number for multiplicaion table"))
for i in range (1,11):
    print(num,'*',i,'=',num*i)
```
