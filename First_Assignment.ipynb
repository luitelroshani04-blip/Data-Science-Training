```python
'''Count Vowels in a Sentence
Description:
Write a program that:
Takes a sentence as input
Counts and displays the number of vowels (a, e, i, o, u, case-insensitive)'''

sentence= input("Enter any sentence")
vowel= "aeiou"
vowel_count= 0
for ch in sentence:
    if ch.lower() in vowel:
        vowel_count += 1
print(vowel_count)
        
```
