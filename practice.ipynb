```python
#Write a program with a for loop that goes through numbers from 1 to 20. 
#Use an if-else statement inside the loop to check if each number is even or odd, 
#and print the count of how many even and odd numbers were found.
a = 0
b= 0
for i in range(1,21):
      if i%2==0:
          a += 1
      else:
          b += 1
print("number of even =", a)
print("number of odd =", b)
```

```python
# Write a program that asks the user to enter a word. 
# Use a for loop to check each letter of the word. 
# Count and print how many vowels (a, e, i, o, u) are inside that word.

a = input("Enter a word").lower()
count = 0
for ch in a:
    if ch in ("a","e","i","o","u"):
        count += 1
print(count)
```

```python
'''Create a list of five numbers in your code (for example, [12, 45, 2, 89, 33]). 
Use a for loop and an if statement to find and print the largest number in that list without using the built-in max() function.'''

a = [12, 45, 2,89, 33]
largest = a[0]
for i in a:
    if i>largest:
        largest = i
print (largest)
        
    
    
```
